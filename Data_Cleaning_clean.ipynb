# Data Cleaning

```
# FIX THE DATATYPE
df['posting_date']=pd.to_datetime(df['posting_date'])
df['application_deadline']=pd.to_datetime(df['application_deadline'])
```

```
df['salary_currency'].unique()
```

```
df['job_title'].unique()
```

```
df['employment_type'].unique()
```

```
df['company_size'].unique()
```

```
df=df.drop(columns=['job_description_length'])
```

```
# mapping
experience_map = {
    'EN': 'Entry',
    'MI': 'Mid',
    'SE': 'Senior',
    'EX': 'Executive'
}
df['experience_level'] = df['experience_level'].map(experience_map)
```

```
#mapping
employment_map={
    'CT':'Contract',
    'FT':'Full Time',
    'PT':'Part Time',
    'FL': 'Freelance'
}

df['employment_type']=df['employment_type'].map(employment_map)
```

```
# mapping
company_map={
    "M":"Mid",
    "L":"Large",
    "S":"Small"
}
df['company_size']=df['company_size'].map(company_map)
    
```

```
# remove unwanted space and convert to lower
df['company_location']=(
    df['company_location'].str.strip()
    .str.lower()
)
```

```
df['employee_residence']=(
    df['employee_residence'].str.strip()
    .str.lower()
)
```

# Data Transformation

```
# create separate dataframe for skills

skills_df=df[['job_id','required_skills']].copy() # required rows only 

skills_df['required_skills']=skills_df['required_skills'].str.split(",") # separated using comma as delimiter

# convert each skill into separate row
skills_df=skills_df.explode('required_skills')

# remove extra space
skills_df['required_skills']=skills_df['required_skills'].str.strip()

# Rename the column 
skills_df=skills_df.rename(columns={'required_skills':'skill'})

skills_df.head(10)
```

```
# create separata datafram for main dataset 

cleaned_jobs_df=df.drop(columns=['required_skills'])

cleaned_jobs_df.head()
```

# Export CSV Files

```
# main dataset 
cleaned_jobs_df.to_csv("cleaned_jobs.csv",index=False)

# skills dataset 
skills_df.to_csv("skills_job.csv",index=False)

print("Successfully exported to CSV Files")
```
